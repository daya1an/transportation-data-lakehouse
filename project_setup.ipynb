## Databricks Widgets

* **Widgets** allow you to pass dynamic values into a notebook.
* Useful for running the **same code across different environments**, such as `dev`, `beta`, and `prod`.
* Example use case: dynamically changing the **catalog name**.

```python
dbutils.widgets.text("catalog", "default_catalog")

catalog = dbutils.widgets.get("catalog")
print(catalog)
```

* Whatever value is provided through the widget is retrieved using `dbutils.widgets.get()`.
* You can then reuse that value throughout the notebook.
